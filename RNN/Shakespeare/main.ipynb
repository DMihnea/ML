```python
# Loading the dataset
import requests
from datasets import Dataset, DatasetDict
import torch

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
text = requests.get(url).text.lower()
print("Text length is", len(text))

# Split the text into chunks of 200 characters
chunk_size = 100
chunks = [
    text[i : i + chunk_size] for i in range(0, len(text) - chunk_size, chunk_size)
]

# Create a dataset from the chunks
full_ds = Dataset.from_dict({"text": chunks})

# Create a train/test split
split = full_ds.train_test_split(test_size=0.1, seed=42)

# Wrap as DatasetDict (optional but nice)
ds = DatasetDict({"train": split["train"], "test": split["test"]})

# Create a vocabulary
vocab = sorted(set(text))
vocab_size = len(vocab)

# Create mappings from characters to integers and vice versa
itos = {i: ch for i, ch in enumerate(vocab)}
stoi = {ch: i for i, ch in enumerate(vocab)}

encoded_train_chunks = [
    torch.tensor([stoi[c] for c in chunk], dtype=torch.long)
    for chunk in ds["train"]["text"]
]

seq_size = 5

# Creating training data 
X_parts = []
y_parts = []


for chunk in encoded_train_chunks:
    windows = chunk.unfold(0, seq_size + 1, 1)
    X_parts.append(windows[:, :seq_size])
    y_parts.append(windows[:, -1])

X = torch.cat(X_parts, dim=0)
y = torch.cat(y_parts, dim=0)
```
